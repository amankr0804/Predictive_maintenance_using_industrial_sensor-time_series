```python
# ============================================================
# TRANSFORMER FOR PREDICTIVE MAINTENANCE / RUL PREDICTION
# NASA C-MAPSS FD001 - FD004
# ============================================================

# ============================================================
# 1. IMPORTS
# ============================================================

import os
import math
import random
import copy
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn

from torch.utils.data import Dataset, DataLoader

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# ============================================================
# 2. CONFIGURATION
# ============================================================

SEED = 42

SEQ_LEN = 30
RUL_CAP = 125

BATCH_SIZE = 256
EPOCHS = 20

LEARNING_RATE = 0.001
WEIGHT_DECAY = 1e-4

# IMPORTANT:
# These are DATASET IDs, NOT file paths.
DATASETS = [
    "FD001",
    "FD002",
    "FD003",
    "FD004"
]

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", DEVICE)

# ============================================================
# 3. REPRODUCIBILITY
# ============================================================

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

# ============================================================
# 4. COLUMN NAMES
# ============================================================

COLUMNS = [
    "unit",
    "cycle",
    "op1",
    "op2",
    "op3"
]

COLUMNS += [
    f"sensor_{i}"
    for i in range(1, 22)
]

print("Number of columns:", len(COLUMNS))

# Should be 26
assert len(COLUMNS) == 26

# ============================================================
# 5. LOAD ONE DATASET
# ============================================================

def load_dataset(dataset):

    # Safety check
    if dataset not in DATASETS:
        raise ValueError(
            f"Invalid dataset: {dataset}"
        )

    train_path = f"/content/train_{dataset}.txt"
    test_path = f"/content/test_{dataset}.txt"
    rul_path = f"/content/RUL_{dataset}.txt"

    print("\nLoading:", dataset)

    print("Train:", train_path)
    print("Test :", test_path)
    print("RUL  :", rul_path)

    # Check files
    if not os.path.exists(train_path):
        raise FileNotFoundError(
            f"Training file not found:\n{train_path}"
        )

    if not os.path.exists(test_path):
        raise FileNotFoundError(
            f"Test file not found:\n{test_path}"
        )

    if not os.path.exists(rul_path):
        raise FileNotFoundError(
            f"RUL file not found:\n{rul_path}"
        )

    # Load training data
    train = pd.read_csv(
        train_path,
        sep=r"\s+",
        header=None,
        names=COLUMNS
    )

    # Load test data
    test = pd.read_csv(
        test_path,
        sep=r"\s+",
        header=None,
        names=COLUMNS
    )

    # Load RUL
    rul = np.loadtxt(rul_path)

    # Make sure RUL is 1D
    rul = np.asarray(rul).reshape(-1)

    print("Train shape:", train.shape)
    print("Test shape :", test.shape)
    print("RUL shape  :", rul.shape)

    return train, test, rul


# ============================================================
# 6. REMOVE CONSTANT FEATURES
# ============================================================

def select_features(train):

    candidate_features = COLUMNS[2:]

    features = []

    for column in candidate_features:

        if train[column].nunique() > 1:
            features.append(column)

    return features


# ============================================================
# 7. ADD RUL TO TRAINING DATA
# ============================================================

def add_rul(train):

    train = train.copy()

    # Maximum cycle for each engine
    max_cycles = (
        train
        .groupby("unit")["cycle"]
        .max()
    )

    # RUL = maximum cycle - current cycle
    train["RUL"] = (
        train["unit"].map(max_cycles)
        - train["cycle"]
    )

    # Cap RUL
    train["RUL"] = train["RUL"].clip(
        upper=RUL_CAP
    )

    return train


# ============================================================
# 8. DATASET CLASS
# ============================================================

class RULDataset(Dataset):

    def __init__(self, X, y):

        self.X = torch.tensor(
            X,
            dtype=torch.float32
        )

        self.y = torch.tensor(
            y,
            dtype=torch.float32
        )

    def __len__(self):

        return len(self.y)

    def __getitem__(self, index):

        return (
            self.X[index],
            self.y[index]
        )


# ============================================================
# 9. CREATE TRAINING WINDOWS
# ============================================================

def create_train_windows(
    data,
    features,
    sequence_length=30
):

    X = []
    y = []

    for unit_id, group in data.groupby("unit"):

        group = group.sort_values("cycle")

        values = (
            group[features]
            .values
            .astype(np.float32)
        )

        rul = (
            group["RUL"]
            .values
            .astype(np.float32)
        )

        # Skip engines shorter than sequence
        if len(values) < sequence_length:
            continue

        for i in range(
            sequence_length,
            len(values) + 1
        ):

            X.append(
                values[
                    i - sequence_length:i
                ]
            )

            y.append(
                rul[i - 1]
            )

    X = np.asarray(
        X,
        dtype=np.float32
    )

    y = np.asarray(
        y,
        dtype=np.float32
    )

    return X, y


# ============================================================
# 10. TRAIN / VALIDATION SPLIT BY ENGINE
# ============================================================

def split_by_engine(
    data,
    validation_ratio=0.2
):

    units = data["unit"].unique()

    units = np.array(units)

    np.random.shuffle(units)

    split = int(
        len(units) *
        (1 - validation_ratio)
    )

    train_units = units[:split]
    val_units = units[split:]

    train_data = data[
        data["unit"].isin(train_units)
    ].copy()

    val_data = data[
        data["unit"].isin(val_units)
    ].copy()

    return train_data, val_data


# ============================================================
# 11. POSITIONAL ENCODING
# ============================================================

class PositionalEncoding(nn.Module):

    def __init__(
        self,
        d_model,
        max_len=500
    ):

        super().__init__()

        position = torch.arange(
            max_len
        ).unsqueeze(1).float()

        div_term = torch.exp(
            torch.arange(
                0,
                d_model,
                2
            ).float()
            *
            (
                -math.log(10000.0)
                / d_model
            )
        )

        pe = torch.zeros(
            max_len,
            d_model
        )

        pe[:, 0::2] = torch.sin(
            position * div_term
        )

        pe[:, 1::2] = torch.cos(
            position * div_term
        )

        pe = pe.unsqueeze(0)

        self.register_buffer(
            "pe",
            pe
        )

    def forward(self, x):

        return (
            x +
            self.pe[
                :, :x.size(1)
            ]
        )


# ============================================================
# 12. TRANSFORMER MODEL
# ============================================================

class TransformerRUL(nn.Module):

    def __init__(
        self,
        input_features,
        d_model=64,
        n_heads=4,
        num_layers=3,
        dropout=0.1
    ):

        super().__init__()

        # Input projection
        self.input_projection = nn.Linear(
            input_features,
            d_model
        )

        # Positional encoding
        self.positional_encoding = PositionalEncoding(
            d_model,
            max_len=SEQ_LEN
        )

        # Transformer encoder
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=n_heads,
            dim_feedforward=d_model * 4,
            dropout=dropout,
            activation="gelu",
            batch_first=True,
            norm_first=True
        )

        self.transformer = nn.TransformerEncoder(
            encoder_layer,
            num_layers=num_layers
        )

        # Normalization
        self.norm = nn.LayerNorm(
            d_model
        )

        # Regression head
        self.regressor = nn.Sequential(

            nn.Linear(
                d_model,
                64
            ),

            nn.GELU(),

            nn.Dropout(
                dropout
            ),

            nn.Linear(
                64,
                32
            ),

            nn.GELU(),

            nn.Linear(
                32,
                1
            )
        )

    def forward(self, x):

        # x:
        # [batch, sequence, features]

        x = self.input_projection(x)

        x = self.positional_encoding(x)

        x = self.transformer(x)

        # Last timestep
        x = x[:, -1, :]

        x = self.norm(x)

        x = self.regressor(x)

        return x.squeeze(-1)


# ============================================================
# 13. TRAIN MODEL
# ============================================================

def train_model(
    model,
    train_loader,
    val_loader
):

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LEARNING_RATE,
        weight_decay=WEIGHT_DECAY
    )

    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode="min",
        factor=0.5,
        patience=2
    )

    criterion = nn.SmoothL1Loss()

    best_val_mae = float("inf")

    best_state = None

    train_losses = []
    val_maes = []

    for epoch in range(EPOCHS):

        # ====================================================
        # TRAIN
        # ====================================================

        model.train()

        total_loss = 0.0

        for X, y in train_loader:

            X = X.to(DEVICE)
            y = y.to(DEVICE)

            optimizer.zero_grad()

            prediction = model(X)

            loss = criterion(
                prediction,
                y
            )

            loss.backward()

            torch.nn.utils.clip_grad_norm_(
                model.parameters(),
                1.0
            )

            optimizer.step()

            total_loss += (
                loss.item() * len(y)
            )

        train_loss = (
            total_loss /
            len(train_loader.dataset)
        )

        # ====================================================
        # VALIDATION
        # ====================================================

        model.eval()

        predictions = []
        actual = []

        with torch.no_grad():

            for X, y in val_loader:

                X = X.to(DEVICE)

                pred = model(X)

                predictions.extend(
                    pred.cpu().numpy()
                )

                actual.extend(
                    y.numpy()
                )

        predictions = np.asarray(
            predictions
        )

        actual = np.asarray(
            actual
        )

        val_mae = mean_absolute_error(
            actual,
            predictions
        )

        val_rmse = np.sqrt(
            mean_squared_error(
                actual,
                predictions
            )
        )

        scheduler.step(val_mae)

        train_losses.append(
            train_loss
        )

        val_maes.append(
            val_mae
        )

        current_lr = optimizer.param_groups[0]["lr"]

        print(
            f"Epoch {epoch + 1:02d}/{EPOCHS} | "
            f"Loss: {train_loss:.4f} | "
            f"Val MAE: {val_mae:.2f} | "
            f"Val RMSE: {val_rmse:.2f} | "
            f"LR: {current_lr:.6f}"
        )

        # ====================================================
        # SAVE BEST MODEL
        # ====================================================

        if val_mae < best_val_mae:

            best_val_mae = val_mae

            best_state = copy.deepcopy(
                model.state_dict()
            )

    # Restore best model
    if best_state is not None:

        model.load_state_dict(
            best_state
        )

    return (
        model,
        train_losses,
        val_maes
    )


# ============================================================
# 14. CREATE TEST WINDOWS
# ============================================================

def create_test_windows(
    test,
    features,
    sequence_length=30
):

    X = []
    units = []

    for unit_id, group in test.groupby("unit"):

        group = group.sort_values(
            "cycle"
        )

        values = (
            group[features]
            .values
            .astype(np.float32)
        )

        # Pad short trajectories
        if len(values) < sequence_length:

            padding = np.repeat(
                values[0:1],
                sequence_length - len(values),
                axis=0
            )

            values = np.vstack(
                [
                    padding,
                    values
                ]
            )

        X.append(
            values[-sequence_length:]
        )

        units.append(
            unit_id
        )

    X = np.asarray(
        X,
        dtype=np.float32
    )

    units = np.asarray(
        units
    )

    return X, units


# ============================================================
# 15. EVALUATE MODEL
# ============================================================

def evaluate_model(
    model,
    X_test,
    true_rul
):

    model.eval()

    X_test_tensor = torch.tensor(
        X_test,
        dtype=torch.float32
    ).to(DEVICE)

    with torch.no_grad():

        predictions = model(
            X_test_tensor
        )

    predictions = (
        predictions
        .cpu()
        .numpy()
    )

    # RUL cannot be negative
    predictions = np.maximum(
        predictions,
        0
    )

    true_rul = np.asarray(
        true_rul
    ).reshape(-1)

    # Safety check
    if len(predictions) != len(true_rul):

        raise ValueError(
            f"Prediction/RUL mismatch: "
            f"{len(predictions)} predictions vs "
            f"{len(true_rul)} true RUL values."
        )

    mae = mean_absolute_error(
        true_rul,
        predictions
    )

    rmse = np.sqrt(
        mean_squared_error(
            true_rul,
            predictions
        )
    )

    r2 = r2_score(
        true_rul,
        predictions
    )

    return (
        predictions,
        mae,
        rmse,
        r2
    )


# ============================================================
# 16. PLOT TRAINING HISTORY
# ============================================================

def plot_training_history(
    train_losses,
    val_maes,
    dataset
):

    plt.figure(
        figsize=(10, 5)
    )

    plt.plot(
        train_losses,
        label="Training Loss"
    )

    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.title(
        f"Training Loss - {dataset}"
    )

    plt.legend()
    plt.grid(alpha=0.3)
    plt.tight_layout()

    plt.savefig(
        f"{dataset}_training_loss.png",
        dpi=300
    )

    plt.show()

    # Validation MAE
    plt.figure(
        figsize=(10, 5)
    )

    plt.plot(
        val_maes,
        label="Validation MAE"
    )

    plt.xlabel("Epoch")
    plt.ylabel("MAE")
    plt.title(
        f"Validation MAE - {dataset}"
    )

    plt.legend()
    plt.grid(alpha=0.3)
    plt.tight_layout()

    plt.savefig(
        f"{dataset}_validation_mae.png",
        dpi=300
    )

    plt.show()


# ============================================================
# 17. PLOT PREDICTIONS
# ============================================================

def plot_results(
    true_rul,
    predictions,
    dataset
):

    # Actual vs predicted by engine
    plt.figure(
        figsize=(12, 5)
    )

    plt.plot(
        true_rul,
        label="Actual RUL"
    )

    plt.plot(
        predictions,
        label="Predicted RUL"
    )

    plt.xlabel("Engine")
    plt.ylabel("Remaining Useful Life")

    plt.title(
        f"Transformer RUL Prediction - {dataset}"
    )

    plt.legend()
    plt.grid(alpha=0.3)
    plt.tight_layout()

    plt.savefig(
        f"{dataset}_prediction.png",
        dpi=300
    )

    plt.show()

    # Scatter plot
    plt.figure(
        figsize=(7, 7)
    )

    plt.scatter(
        true_rul,
        predictions,
        alpha=0.7
    )

    minimum = min(
        true_rul.min(),
        predictions.min()
    )

    maximum = max(
        true_rul.max(),
        predictions.max()
    )

    plt.plot(
        [minimum, maximum],
        [minimum, maximum],
        linestyle="--"
    )

    plt.xlabel("Actual RUL")
    plt.ylabel("Predicted RUL")

    plt.title(
        f"Actual vs Predicted RUL - {dataset}"
    )

    plt.grid(alpha=0.3)
    plt.tight_layout()

    plt.savefig(
        f"{dataset}_scatter.png",
        dpi=300
    )

    plt.show()


# ============================================================
# 18. RUN ONE DATASET
# ============================================================

def run_dataset(dataset):

    print("\n")
    print("=" * 70)
    print(f"TRAINING TRANSFORMER ON {dataset}")
    print("=" * 70)

    # ========================================================
    # LOAD
    # ========================================================

    train, test, rul = load_dataset(
        dataset
    )

    # True RUL for test
    true_rul = np.asarray(
        rul
    ).reshape(-1)

    # ========================================================
    # CHECK TEST ENGINES
    # ========================================================

    test_units = (
        test["unit"]
        .unique()
    )

    print(
        "Test engines:",
        len(test_units)
    )

    print(
        "True RUL values:",
        len(true_rul)
    )

    if len(test_units) != len(true_rul):

        raise ValueError(
            f"{dataset}: Number of test engines "
            f"({len(test_units)}) does not match "
            f"RUL values ({len(true_rul)})."
        )

    # ========================================================
    # SELECT FEATURES
    # ========================================================

    features = select_features(
        train
    )

    print(
        "\nNumber of selected features:",
        len(features)
    )

    print(
        "Features:",
        features
    )

    # ========================================================
    # ADD RUL
    # ========================================================

    train = add_rul(
        train
    )

    print(
        "\nRUL range:",
        train["RUL"].min(),
        "to",
        train["RUL"].max()
    )

    # ========================================================
    # TRAIN / VALIDATION SPLIT
    # ========================================================

    train_data, val_data = split_by_engine(
        train,
        validation_ratio=0.2
    )

    print(
        "Training engines:",
        train_data["unit"].nunique()
    )

    print(
        "Validation engines:",
        val_data["unit"].nunique()
    )

    # ========================================================
    # SCALING
    # ========================================================

    scaler = StandardScaler()

    # Fit ONLY on training engines
    scaler.fit(
        train_data[features]
    )

    train_data = train_data.copy()
    val_data = val_data.copy()
    test = test.copy()

    train_data.loc[:, features] = (
        scaler.transform(
            train_data[features]
        )
    )

    val_data.loc[:, features] = (
        scaler.transform(
            val_data[features]
        )
    )

    test.loc[:, features] = (
        scaler.transform(
            test[features]
        )
    )

    # ========================================================
    # CREATE TRAINING WINDOWS
    # ========================================================

    X_train, y_train = create_train_windows(
        train_data,
        features,
        SEQ_LEN
    )

    X_val, y_val = create_train_windows(
        val_data,
        features,
        SEQ_LEN
    )

    print(
        "\nTraining windows:",
        X_train.shape
    )

    print(
        "Validation windows:",
        X_val.shape
    )

    if len(X_train) == 0:

        raise ValueError(
            f"{dataset}: No training windows created."
        )

    if len(X_val) == 0:

        raise ValueError(
            f"{dataset}: No validation windows created."
        )

    # ========================================================
    # DATASETS
    # ========================================================

    train_dataset = RULDataset(
        X_train,
        y_train
    )

    val_dataset = RULDataset(
        X_val,
        y_val
    )

    # pin_memory only useful with CUDA
    use_pin_memory = torch.cuda.is_available()

    train_loader = DataLoader(
        train_dataset,
        batch_size=BATCH_SIZE,
        shuffle=True,
        pin_memory=use_pin_memory
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        pin_memory=use_pin_memory
    )

    # ========================================================
    # MODEL
    # ========================================================

    model = TransformerRUL(
        input_features=len(features),
        d_model=64,
        n_heads=4,
        num_layers=3,
        dropout=0.1
    )

    model = model.to(
        DEVICE
    )

    print(
        "\nModel created successfully."
    )

    # ========================================================
    # TRAIN
    # ========================================================

    model, train_losses, val_maes = train_model(
        model,
        train_loader,
        val_loader
    )

    # ========================================================
    # TRAINING PLOTS
    # ========================================================

    plot_training_history(
        train_losses,
        val_maes,
        dataset
    )

    # ========================================================
    # TEST WINDOWS
    # ========================================================

    X_test, test_units = create_test_windows(
        test,
        features,
        SEQ_LEN
    )

    print(
        "\nTest windows:",
        X_test.shape
    )

    # ========================================================
    # TEST
    # ========================================================

    predictions, mae, rmse, r2 = evaluate_model(
        model,
        X_test,
        true_rul
    )

    # ========================================================
    # RESULTS
    # ========================================================

    print("\n")
    print("=" * 50)
    print(f"{dataset} RESULTS")
    print("=" * 50)

    print(
        f"MAE  : {mae:.2f} cycles"
    )

    print(
        f"RMSE : {rmse:.2f} cycles"
    )

    print(
        f"R2   : {r2:.4f}"
    )

    # ========================================================
    # SAVE PREDICTIONS
    # ========================================================

    prediction_df = pd.DataFrame({

        "unit": test_units,

        "actual_RUL": true_rul,

        "predicted_RUL": predictions

    })

    prediction_file = (
        f"{dataset}_transformer_predictions.csv"
    )

    prediction_df.to_csv(
        prediction_file,
        index=False
    )

    print(
        "Predictions saved:",
        prediction_file
    )

    # ========================================================
    # SAVE MODEL
    # ========================================================

    model_file = (
        f"transformer_{dataset}.pth"
    )

    torch.save({

        "model_state_dict":
            model.state_dict(),

        "features":
            features,

        "sequence_length":
            SEQ_LEN,

        "rul_cap":
            RUL_CAP,

        "scaler_mean":
            scaler.mean_,

        "scaler_scale":
            scaler.scale_

    }, model_file)

    print(
        "Model saved:",
        model_file
    )

    # ========================================================
    # PLOTS
    # ========================================================

    plot_results(
        true_rul,
        predictions,
        dataset
    )

    # ========================================================
    # RETURN RESULTS
    # ========================================================

    return {

        "Dataset":
            dataset,

        "Engines":
            len(true_rul),

        "Features":
            len(features),

        "MAE":
            mae,

        "RMSE":
            rmse,

        "R2":
            r2
    }


# ============================================================
# 19. RUN ALL FOUR DATASETS
# ============================================================

all_results = []

for dataset in DATASETS:

    result = run_dataset(
        dataset
    )

    all_results.append(
        result
    )


# ============================================================
# 20. FINAL COMPARISON
# ============================================================

results_df = pd.DataFrame(
    all_results
)

print("\n")
print("=" * 70)
print("FINAL TRANSFORMER COMPARISON")
print("=" * 70)

print(
    results_df.to_string(
        index=False
    )
)


# ============================================================
# 21. SAVE FINAL RESULTS
# ============================================================

results_df.to_csv(
    "transformer_results_all.csv",
    index=False
)

print(
    "\nResults saved to:"
)

print(
    "transformer_results_all.csv"
)
```
